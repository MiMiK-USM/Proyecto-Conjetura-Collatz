```python
def secuencia_collatz(n):
    trayectoria = [n]
    while n > 1:
        if n % 2 == 0:
            n = n//2
        else:
            n = 3 * n + 1
            trayectoria.append(n)
    return trayectoria
print(secuencia_collatz(27))
```

```python
import matplotlib.pyplot as plt
datos = secuencias_collatz(27)
plt.plot(datos)
plt.title("Trayectoria de Collatz para n=27")
plt.xlabel("Pasos")
plt.xlabel("Valor")
plt.show()
```
