```python
import matplotlib.pyplot as plt
import numpy as np

# Reutilizamos las funciones base de Collatz
def secuencia_collatz(n):
    trayectoria = [n]
    while n > 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        trayectoria.append(n)
    return trayectoria

def calcular_metricas_collatz(N):
    tiempos_vuelo = []
    altitudes_maximas = []
    for i in range(1, N + 1):
        trayectoria = secuencia_collatz(i)
        tiempos_vuelo.append(len(trayectoria) - 1)
        altitudes_maximas.append(max(trayectoria))
    return np.array(tiempos_vuelo), np.array(altitudes_maximas)

# 1. Análisis estadístico escalado a N = 100,000 números
N_grande = 100000
n_valores = np.arange(1, N_grande + 1)
tiempos, altitudes = calcular_metricas_collatz(N_grande)

print(f"--- Estadísticas de T(n) para N={N_grande} ---")
print(f"Media: {np.mean(tiempos):.2f}")
print(f"Mediana: {np.median(tiempos)}")
print(f"Desviación Estándar: {np.std(tiempos):.2f}")
print(f"Percentil 95: {np.percentile(tiempos, 95)}")
```

```python
# 2. Estructura por paridad (Boxplots de Pares vs Impares)
tiempos_pares = tiempos[n_valores % 2 == 0]
tiempos_impares = tiempos[n_valores % 2 != 0]

plt.figure(figsize=(8, 6))
plt.boxplot([tiempos_pares, tiempos_impares], labels=['Pares', 'Impares'])
plt.title('Distribución de T(n) según Paridad del número inicial')
plt.ylabel('Tiempo de vuelo $T(n)$')
plt.grid(True, alpha=0.3)
plt.savefig('../figs/collatz_paridad_boxplot.png', dpi=300)
plt.show()
```

```python
# 3. Encontrar récords sucesivos (T(n) > T(k) para todo k < n)
records_n = []
records_t = []
max_tiempo_previo = -1

for n in range(1, N_grande + 1):
    t_actual = tiempos[n - 1]
    if t_actual > max_tiempo_previo:
        records_n.append(n)
        records_t.append(t_actual)
        max_tiempo_previo = t_actual
        
print("--- Primeros Récords Sucesivos Encontrados ---")
for idx, (rn, rt) in enumerate(zip(records_n[:20], records_t[:20])):
    print(f"Récord {idx+1}: n = {rn} tiene un tiempo de vuelo de {rt} pasos")

# Gráfico de crecimiento de récords
plt.figure(figsize=(10, 5))
plt.plot(records_n, records_t, marker='o', linestyle='-', color='red')
plt.xscale('log')
plt.title('Crecimiento de Récords del Tiempo de Vuelo')
plt.xlabel('Número inicial $n$ (Escala Logarítmica)')
plt.ylabel('Tiempo de vuelo récord $T(n)$')
plt.grid(True, which="both", alpha=0.3)
plt.savefig('../figs/collatz_records.png', dpi=300)
plt.show()
```
