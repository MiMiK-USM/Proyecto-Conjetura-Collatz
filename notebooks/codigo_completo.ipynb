```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def secuencia_collatz(n):
    trayectoria = [n]
    while n > 1:
        if n % 2 == 0:
            n = n//2
        else:
            n = 3 * n + 1
            trayectoria.append(n)
    return trayectoria
print(secuencia_collatz(27))
```

```python
tiempos_vuelo = []
altitud_maxima = []
for i in range(1, 10001):
    t = secuencia_collatz(i)
    tiempos_vuelo.append(len(t) - 1)
    altitud_maxima.append(max(t))
```

```python
datos = secuencia_collatz(27)
plt.plot(datos)
plt.title("Trayectoria de Collatz para n=27")
plt.xlabel("Pasos")
plt.ylabel("Valor")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(tiempos_vuelo, bins=100, color='orange', edgecolor='black')
plt.title("Histograma de Tiempos de Vuelo (n=1 hasta 10.000)")
plt.xlabel("Pasos para llegar a 1")
plt.ylabel("Cantidad de números")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
numeros = list(range(1, 10001)) 
plt.scatter(numeros, tiempos_vuelo, s=1, alpha=0.5)
plt.title("Dispersión: Número inicial vs Tiempo de Vuelo")
plt.xlabel("Número inicial (n)")
plt.ylabel("Tiempo de Vuelo (pasos)")
plt.show()
```

```python
resultados = list(zip(range(1, 10001), tiempos_vuelo))
top_10 = sorted(resultados, key=lambda x: x[1], reverse=True)[:10]

print("--- TOP 10 NÚMEROS CON MAYOR TIEMPO DE VUELO ---")
for n, tiempo in top_10:
    print(f"Número: {n} -> Tiempo de vuelo: {tiempo} pasos")
```

```python
def info_collatz(n):
    t = 0
    a_max = n
    valor_actual = n
    while valor_actual > 1:
        if valor_actual % 2 == 0:
            valor_actual //= 2
        else:
            valor_actual = 3 * valor_actual + 1
        t += 1
        if valor_actual > a_max:
            a_max = valor_actual
    return t, a_max

n_final = 100000
datos = []

for i in range(1, n_final + 1):
    tiempo, altitud = info_collatz(i)
    datos.append([i, tiempo, altitud])

df = pd.DataFrame(datos, columns=['n (La Manzana)', 'T(n) (Pasos que dio)', 'A(n) (Altura que alcanzó)'])

df.to_csv('datos_collatz_100k.csv', index=False)

print("¡Archivo 'datos_collatz_100k.csv' creado con éxito!")
```
