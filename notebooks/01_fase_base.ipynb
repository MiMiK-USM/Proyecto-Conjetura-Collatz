```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Definición de la función de Collatz
def secuencia_collatz(n):
    trayectoria = [n]
    while n > 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        trayectoria.append(n)
    return trayectoria

# 2. Función para obtener las métricas de un rango
def calcular_metricas_collatz(N):
    tiempos_vuelo = []
    altitudes_maximas = []
    for i in range(1, N + 1):
        trayectoria = secuencia_collatz(i)
        tiempos_vuelo.append(len(trayectoria) - 1)
        altitudes_maximas.append(max(trayectoria))
    return np.array(tiempos_vuelo), np.array(altitudes_maximas)

# Calcular métricas para N=10000
n_valores = np.arange(1, 10001)
tiempos, altitudes = calcular_metricas_collatz(10000)
print("¡Estructura base lista! N=10000 calculado.")
```

```python
# Gráfico 1: Diagrama de dispersión T(n) vs n
plt.figure(figsize=(10, 6))
plt.scatter(n_valores, tiempos, alpha=0.3, s=1, color='blue')
plt.title('Tiempo de vuelo $T(n)$ vs $n$')
plt.xlabel('$n$')
plt.ylabel('Tiempo de vuelo $T(n)$')
plt.grid(True, alpha=0.3)
plt.savefig('../figs/collatz_dispersion.png', dpi=300)
plt.show()
```

```python
# Gráfico 2: Histograma de T(n)
plt.figure(figsize=(10, 6))
plt.hist(tiempos, bins=50, color='purple', edgecolor='black', alpha=0.7)
plt.title('Histograma de Tiempos de Vuelo $T(n)$')
plt.xlabel('Tiempo de vuelo $T(n)$')
plt.ylabel('Frecuencia')
plt.grid(True, alpha=0.3)
plt.savefig('../figs/collatz_histograma.png', dpi=300)
plt.show()
```
